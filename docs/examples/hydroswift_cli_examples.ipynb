**`search` -> `get` -> `show` / `merge` / `plot`** (run `hyswift` without arguments for guided interactive mode).

## 1. Version Check & Help

````python
!hyswift --version
!hyswift --help
````

## 2. Find: `hyswift search`
Search across NWDP, CWC Flood Forecasting, and CWC RSMS. Results can be saved to a catalog JSON file.

````python
# Search for rainfall in Karnataka and save to cat.json
!hyswift search rainfall --where Karnataka --period 2020:2022 --save cat.json
````

````python
# Search river discharge stations for Krishna basin from CWC
!hyswift search discharge --basin Krishna --sources CWC
````

## 3. Get: `hyswift get`
Download specific rows of a saved catalog table with audit logging in `manifest.json`.

````python
# Fetch rows 0 and 1 from the saved catalog
!hyswift get --from cat.json --rows 0,1 --period 2020:2022 --out output
````

## 4. Use: `hyswift show`
Display a tidy summary of downloaded datasets.

````python
!hyswift show output
````

## 5. Export: `hyswift merge`
Merge downloaded records into a single consolidated file (`.parquet`, `.csv`, or `.gpkg`).

````python
!hyswift merge output output/merged.parquet
````

## 6. Feature-Rich Hydrological Plotting: `hyswift plot`

Generate instant hydrological figures across multiple analysis kinds (`timeseries`, `hydrograph`, `hyetograph`, `rating_curve`, `compare`, `fdc`):

````python
# Standard hydrographs with CWC flood alert thresholds (Warning, Danger, HFL)
!hyswift plot output output/plots --kind timeseries

# Dual-axis hydrograph combining stage and discharge
# !hyswift plot output output/plots --kind hydrograph

# Empirical hydraulic rating curve (H-Q scatter + fitted curve)
# !hyswift plot output output/plots --kind rating_curve
````

## 7. Interactive Geospatial Mapping: `hyswift map`

Generate self-contained Leaflet HTML maps with marker clustering, multi-basemap layer switching, and flood alert highlights directly from search catalogs or output folders:

````python
# Map stations from search catalog or downloaded output
!hyswift map output --out output/station_map.html --basemap carto-light
````

## 8. Reservoirs: `hyswift rsms`
CWC Reservoir Storage Monitoring System tools.

````python
# Fetch latest daily storage snapshot
!hyswift rsms daily --out output/cwc_rsms/daily_snapshot.csv

# Backfill historical weekly bulletins
!hyswift rsms backfill --start 2025-05-15 --out output/cwc_rsms/backfill.csv
````

## 9. Utilities: `cite` & `coffee`
