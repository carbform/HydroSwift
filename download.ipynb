# HydroSwift ⚡ Quick Start: Find -> Review -> Get -> Use

HydroSwift provides automated retrieval, processing, and tidy standardization of Indian hydrological observations.

Supported data sources:
* **NWDP** (National Water Data Portal - CKAN open data)
* **CWC FFS** (Central Water Commission Flood Forecasting System: stage & rating-curve discharge)
* **CWC RSMS** (Reservoir Storage Monitoring System: 180+ major reservoirs)
* **India-WRIS REST API** (Official OpenAPI catalog)


```python
import hydroswift as hs
print(f"HydroSwift version: {hs.__version__} ({hs.__codename__})")
```

## 1. Find (Discovery)
Search for datasets by variable, geographic location, or basin across all active sources.

```python
# Search for rainfall observations in Karnataka between 2020 and 2022
cat = hs.search("rainfall", location="Karnataka", period=("2020", "2022"))
cat
```

## 2. Review (Provenance & Estimates)
Check what each source reported, identify any truncated listings or dropped resources, and estimate data volume.

```python
# Source status metadata
cat.attrs.get("source_status")
```

```python
# Estimate records before downloading
hs.estimate(cat.head(3))
```

## 3. Get (Fetch)
Download the selected datasets with progress tracking and automatic  generation.

```python
# Fetch selected catalog rows into the output directory
run = hs.fetch(cat.head(2), output_dir="output", period=("2020", "2022"))
run
```

## 4. Use (Tidy Long DataFrame)
Load downloaded observations into one standardized long table (, , , , , , , , , , ).

```python
tidy_df = hs.load(run)
tidy_df.head(10)
```

## 5. Export & Visualization

```python
# Export to compressed Parquet or CSV
hs.merge(run, output="output/rainfall_karnataka.parquet")

# Plot time series figures for each station
plot_files = hs.plot(run, output_dir="output/plots")
print(f"Generated {len(plot_files)} plot figures.")
```

## 6. CWC Flood Forecasting & Reservoirs (RSMS)

```python
# Search river discharge forecasting stations for Krishna basin
cwc_stations = hs.search("discharge", basin="Krishna", sources=["CWC"])
cwc_stations.head()
```

```python
# Real-time daily storage snapshot of 180+ major national reservoirs
res_snapshot = hs.rsms.daily()
res_snapshot.head()
```
